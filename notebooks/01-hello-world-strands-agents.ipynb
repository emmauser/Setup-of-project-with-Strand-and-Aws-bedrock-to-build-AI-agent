```python
import os
import boto3
from strands import Agent
from strands.models import BedrockModel

print("Import successful")
```

```python
#Define a session to access to Amazon Bedrock
session = boto3.Session(region_name='us-east-1')
```

```python
#Defini what model we use
kimi_model = BedrockModel(
    model_id="moonshotai.kimi-k2.5",  # Change l'ID si différent
    boto_session=session,
    streaming=False
)
```

```python
#Create agent for our model
agent = Agent(model=kimi_model)
response = agent("Quelle est la capitale de la France ?")
print(response)
```

```python
#Display information about our model
print(f"Current Agent model provider : {agent.model} ")
print(f"Current Agent model id : {agent.model.config['model_id']} ")
print("-" *100)
print("\n")
```

```python
print(f"Message : {response.message}")
print("-" * 100 + "\n")

print(f"Metrics : {response.metrics}")
print("-" * 100 + "\n")

print(f"State: {response.state}")
print("-" * 100 + "\n")

print("Stop Reason : {response.stop_reason}")
print("-" * 100 + "\n")
```
