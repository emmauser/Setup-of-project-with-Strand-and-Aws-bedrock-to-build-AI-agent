```python
from strands.models import BedrockModel
import boto3
from strands import Agent


session = boto3.Session(region_name='us-east-1')

kimi_model = BedrockModel(
    model_id="moonshotai.kimi-k2.5",  # Change l'ID si différent
    boto_session=session,
    streaming=False
)

agent = Agent(model=kimi_model)
response = agent("Quelle est la capitale de la France ?")
print(response)
```
